IMPORTING THE DATA

```python
conn = sqlite3.connect('customer_churn.db')


sql_query = """ 
        SELECT name
        FROM sqlite_master
        WHERE type='table';
"""

tables = pd.read_sql(sql_query, conn)

for table_name in tables['name']:
    df = pd.read_sql(f'SELECT * FROM {table_name}', conn)
    globals()[f"df_{table_name}"] = df
    print(f"created DataFrame : df_{table_name}")

conn.close()
```

Now the first data analysis we are going to perform is to calcualte churn rate

```python
df.columns
```

```python
churn_rate = df['churn_flag'].mean()*100
print(f"The churn rate is {churn_rate:.2f}%")
```

now we calculate the retentation rate .

```python
retention_rate = 100- churn_rate
print(f"The retention rate is {retention_rate:.2f}%")
```

Now we'll check the churn by plan type

```python
churn_by_plan = (df.groupby('plan_type')['churn_flag'].mean().mul(100).round(2).reset_index(name = 'churn_rate_pct'))
print(churn_by_plan)
```

Now we'll check the churn by state type + sum of the revenue and count of users

```python
churn_by_state = (df.groupby('state')['churn_flag'].mean().mul(100).round(2).reset_index(name = 'churn_rate_pct'))
print(churn_by_state)
```

Now we'll check the churn by subscription type + sum of the revenue and count of users

```python
churn_by_subscription_type = (df.groupby('subscription_type')['churn_flag'].mean().mul(100).round(2).reset_index(name = 'churn_rate_pct'))
print(churn_by_subscription_type)
```
